```python
import os

file_path = '/Users/sameralzubaidi/Desktop/kids_marketplace_app/p2p-kids-marketplace/src/screens/profile/ProfileScreen.tsx'

try:
    with open(file_path, 'r') as f:
        lines = f.readlines()
        
        # User wants lines 480 to 550 (presumably 1-based)
        # 1-based line 480 is index 479
        # 1-based line 550 is index 549
        
        start_line = 480
        end_line = 550
        
        selected_lines = lines[start_line-1:end_line]
        
        print(f"--- Reading lines {start_line} to {end_line} ---")
        for i, line in enumerate(selected_lines):
            print(f"{start_line + i}: {line}", end='')
            
except FileNotFoundError:
    print(f"File not found: {file_path}")
except Exception as e:
    print(f"Error: {e}")
```
